```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student-scores.csv")

subjects = ["math_score", "physics_score", "chemistry_score", "biology_score", "english_score", "history_score", "geography_score"]

# Compute averages
avg_scores = df.groupby("gender")[subjects].mean()

avg_scores.T.plot(kind="bar")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks of Male vs Female Students")
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

```python
avg_job = df.groupby("part_time_job")[subjects].mean()

avg_job.T.plot(kind="line", marker='o')
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks Variation: Part-time Job vs No Job")
plt.legend(["No Job", "Has Job"])
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["weekly_self_study_hours"], bins=10)
plt.xlabel("Weekly Self Study Hours")
plt.ylabel("Number of Students")
plt.title("Distribution of Self Study Hours")
plt.tight_layout()
plt.show()
```

```python
activities = df["extracurricular_activities"].value_counts()

plt.pie(activities, labels=activities.index, autopct="%1.1f%%")
plt.title("Percentage of Students in Extracurricular Activities")
plt.show()
```
