```python
import pandas as pd
titanic = pd.read_csv('tested.csv')
```

```python
columns_to_delete = []
for column in titanic.columns:
    count = titanic[column].isnull().sum()
    if count > titanic.shape[0] * 0.5:
     columns_to_delete.append(column)

print (columns_to_delete)
```

```python
titanic_cleaned = titanic.drop(columns=columns_to_delete)
```

```python
titanic_cleaned = titanic_cleaned.dropna()
```

```python
titanic_cleaned.to_csv('cleaned.csv',index=False)
```

```python
titanic_replaced = titanic.copy()
if 'Age' in titanic_replaced.columns:
    titanic_replaced['Age'].fillna(titanic_replaced['Age'].mean(),inplace=True)

print(f"{titanic_replaced['Age'].mean():.2f}")
```
